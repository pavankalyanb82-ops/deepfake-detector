# Phase 2: Exploratory Data Analysis (EDA) - Deepfake Image Detection

Welcome to **Phase 2** of the Deepfake Detector project!

### Why is EDA important for Deepfake Detection?
Before building a deep learning model, we must understand our dataset's properties:
1. **Class Balance**: Is our dataset balanced between `Real` and `Fake` images? Imbalanced datasets bias neural networks toward the majority class.
2. **Image Dimensions**: Are images uniform in size and aspect ratio?
3. **Visual GAN Artifacts**: What visual cues differentiate AI-generated faces from real human photos? (e.g., asymmetric pupil shapes, unnatural ear geometry, background warping, texture smudging).

Let's explore our data step-by-step!

```python
import os
import glob
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

DATA_DIR = '../data/raw'
real_paths = glob.glob(os.path.join(DATA_DIR, 'real', '*.*'))
fake_paths = glob.glob(os.path.join(DATA_DIR, 'fake', '*.*'))

print(f'Total Real Images: {len(real_paths)}')
print(f'Total Fake Images: {len(fake_paths)}')
print(f'Total Dataset Size: {len(real_paths) + len(fake_paths)}')
```

## 1. Class Balance & Image Dimension Checks

```python
# Inspect image dimensions and color channels
sample_real = Image.open(real_paths[0])
sample_fake = Image.open(fake_paths[0])

print(f'Real Image Size (W x H): {sample_real.size}, Format: {sample_real.format}, Mode: {sample_real.mode}')
print(f'Fake Image Size (W x H): {sample_fake.size}, Format: {sample_fake.format}, Mode: {sample_fake.mode}')

# Plot class balance
plt.figure(figsize=(6, 4))
plt.bar(['Real', 'Fake'], [len(real_paths), len(fake_paths)], color=['#2ecc71', '#e74c3c'])
plt.ylabel('Image Count')
plt.title('Dataset Class Distribution')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

## 2. Visual Comparison Grid: Real vs. Fake Faces

Below we display a side-by-side grid comparing real faces (top row) with synthetic/fake faces (bottom row).

```python
num_samples = 4
fig, axes = plt.subplots(2, num_samples, figsize=(14, 7))

# Real images (Top Row)
for idx in range(num_samples):
    img = Image.open(real_paths[idx])
    axes[0, idx].imshow(img)
    axes[0, idx].set_title(f'Real #{idx+1}')
    axes[0, idx].axis('off')

# Fake images (Bottom Row)
for idx in range(num_samples):
    img = Image.open(fake_paths[idx])
    axes[1, idx].imshow(img)
    axes[1, idx].set_title(f'Fake #{idx+1}')
    axes[1, idx].axis('off')

plt.suptitle('Real vs. Fake Facial Image Comparison', fontsize=16)
plt.tight_layout()
plt.show()
```

## 3. Human Intuition & Visual Artifact Analysis

When examining GAN-generated faces (e.g. StyleGAN / Deepfakes), keep an eye out for these signature visual artifacts:

- **Pupil & Iris Asymmetry**: GANs frequently generate irises that are not perfectly circular or pupils with mismatched shapes and reflection points.
- **Jewelry & Earring Discrepancies**: AI models struggle with fine symmetry in earrings or eyeglasses; one side might have an earring while the other side fades into the hair.
- **Hair & Background Texture Blur**: Subtle boundary distortions often occur where hair strands blend into the background background context.
- **Skin Over-smoothing**: Generated skin textures can look unnaturally smooth or lack fine pores.

Recognizing these visual patterns helps explain what feature maps our CNN (EfficientNet/ResNet) will learn!
